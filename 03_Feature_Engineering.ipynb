```python
# ==========================================================
# Import Required Libraries
# ==========================================================

import pandas as pd
import numpy as np
import os

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 1000)

print("Libraries imported successfully!")
```

```python
# ==========================================================
# Load Cleaned Data
# ==========================================================

sales_df = pd.read_csv("../data/processed/clean_sales.csv")
calendar_df = pd.read_csv("../data/processed/clean_calendar.csv")
prices_df = pd.read_csv("../data/processed/clean_prices.csv")

print("Datasets loaded successfully!")
```

```python
calendar_df["date"] = pd.to_datetime(calendar_df["date"])
```

```python
# ==========================================================
# Time Features
# ==========================================================

calendar_df["day"] = calendar_df["date"].dt.day
calendar_df["month_name"] = calendar_df["date"].dt.month_name()
calendar_df["quarter"] = calendar_df["date"].dt.quarter
calendar_df["week"] = calendar_df["date"].dt.isocalendar().week.astype(int)

print("Time features created successfully!")
```

```python
calendar_df[
    [
        "date",
        "year",
        "month",
        "month_name",
        "day",
        "week",
        "quarter",
    ]
].head()
```

```python
# ==========================================================
# Weekend Indicator
# ==========================================================

calendar_df["is_weekend"] = calendar_df["weekday"].isin(["Saturday", "Sunday"]).astype(int)

print("Weekend feature created successfully!")
```

```python
calendar_df[["weekday", "wday", "is_weekend"]].head(10)
```

```python
# ==========================================================
# Event Indicator
# ==========================================================

calendar_df["has_event"] = (
    (calendar_df["event_name_1"] != "No_Event") |
    (calendar_df["event_name_2"] != "No_Event")
).astype(int)

print("Event indicator created successfully!")
```

```python
calendar_df[
    [
        "date",
        "event_name_1",
        "event_name_2",
        "has_event"
    ]
].head(15)
```

```python
print("Total Event Days     :", calendar_df["has_event"].sum())
print("Total Non-Event Days :", len(calendar_df) - calendar_df["has_event"].sum())
```

```python
# ==========================================================
# Event Indicator
# ==========================================================

calendar_df["has_event"] = (
    (calendar_df["event_name_1"] != "No_Event") |
    (calendar_df["event_name_2"] != "No_Event")
).astype(int)

print("Event indicator created successfully!")
```

```python
calendar_df[
    [
        "date",
        "event_name_1",
        "event_name_2",
        "has_event"
    ]
].head(15)
```

```python
print("Total Event Days     :", calendar_df["has_event"].sum())
print("Total Non-Event Days :", len(calendar_df) - calendar_df["has_event"].sum())
```

```python
# ==========================================================
# SNAP Indicator
# ==========================================================

calendar_df["snap_active"] = (
    (calendar_df["snap_CA"] == 1) |
    (calendar_df["snap_TX"] == 1) |
    (calendar_df["snap_WI"] == 1)
).astype(int)

print("SNAP feature created successfully!")
```

```python
calendar_df[
    [
        "date",
        "snap_CA",
        "snap_TX",
        "snap_WI",
        "snap_active"
    ]
].head(15)
```

```python
print("SNAP Active Days :", calendar_df["snap_active"].sum())
print("Non SNAP Days    :", len(calendar_df) - calendar_df["snap_active"].sum())
```

```python
feature_path = "../data/features"

os.makedirs(feature_path, exist_ok=True)

print("Feature folder is ready!")
```

```python
calendar_df.to_csv("../data/features/calendar_features.csv", index=False)
sales_df.to_csv("../data/features/sales_features.csv", index=False)
prices_df.to_csv("../data/features/prices_features.csv", index=False)

print("Feature Engineering Completed Successfully!")
```
