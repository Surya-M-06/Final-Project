```python
# ==========================================================
# Import Required Libraries
# ==========================================================

import pandas as pd
import numpy as np
import os

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 1000)

print("Libraries imported successfully!")
```

```python
# ==========================================================
# Load Datasets
# ==========================================================

sales_df = pd.read_csv("../data/raw/sales_train_validation.csv")
calendar_df = pd.read_csv("../data/raw/calendar.csv")
prices_df = pd.read_csv("../data/raw/sell_prices.csv")

print("Datasets loaded successfully!")
```

```python
# ==========================================================
# Dataset Shapes
# ==========================================================

print("Sales Dataset Shape    :", sales_df.shape)
print("Calendar Dataset Shape :", calendar_df.shape)
print("Prices Dataset Shape   :", prices_df.shape)
```

```python
# ==========================================================
# Missing Values
# ==========================================================

print("Sales Dataset")
display(sales_df.isnull().sum())

print("\nCalendar Dataset")
display(calendar_df.isnull().sum())

print("\nPrices Dataset")
display(prices_df.isnull().sum())
```

```python
# ==========================================================
# Convert Date Column
# ==========================================================

calendar_df["date"] = pd.to_datetime(calendar_df["date"])

print("Date column converted successfully!")
print(calendar_df["date"].dtype)
```

```python
# ==========================================================
# Check Duplicate Records
# ==========================================================

print("Sales Dataset Duplicates    :", sales_df.duplicated().sum())
print("Calendar Dataset Duplicates :", calendar_df.duplicated().sum())
print("Prices Dataset Duplicates   :", prices_df.duplicated().sum())
```

```python
# ==========================================================
# Handle Missing Values
# ==========================================================

event_columns = [
    "event_name_1",
    "event_type_1",
    "event_name_2",
    "event_type_2"
]

calendar_df[event_columns] = calendar_df[event_columns].fillna("No_Event")

print("Missing values handled successfully!")
```

```python
# ==========================================================
# Verify Missing Values
# ==========================================================

calendar_df[event_columns].isnull().sum()
```

```python
# ==========================================================
# Check Data Types
# ==========================================================

print(calendar_df.dtypes)
```

```python
# ==========================================================
# Create Processed Data Folder
# ==========================================================

processed_path = "../data/processed"

os.makedirs(processed_path, exist_ok=True)

print("Processed folder is ready!")
```

```python
# ==========================================================
# Save Cleaned Datasets
# ==========================================================

sales_df.to_csv("../data/processed/clean_sales.csv", index=False)
calendar_df.to_csv("../data/processed/clean_calendar.csv", index=False)
prices_df.to_csv("../data/processed/clean_prices.csv", index=False)

print("Cleaned datasets saved successfully!")
```
