Safety Stock=(Maximum Daily Sales×Maximum Lead Time)−(Average Daily Sales×Average Lead Time)
Reorder Point=(Average Daily Sales×Average Lead Time)+Safety Stock

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
sales_df = pd.read_csv("../data/processed/clean_sales.csv")
sales_df.head()
```

```python
daily_sales_columns = [
    col for col in sales_df.columns
    if col.startswith("d_")
]

print("Number of daily sales columns:", len(daily_sales_columns))
print("First 10 columns:", daily_sales_columns[:10])
print("Last 10 columns:", daily_sales_columns[-10:])
```

```python
sales_df["average_daily_sales"] = sales_df[
    daily_sales_columns
].mean(axis=1)

sales_df[
    ["item_id", "store_id", "average_daily_sales"]
].head(10)
```

```python
sales_df["maximum_daily_sales"] = sales_df[
    daily_sales_columns
].max(axis=1)

sales_df[
    ["item_id", "store_id", "maximum_daily_sales"]
].head(10)
```

```python
AVERAGE_LEAD_TIME = 7
MAXIMUM_LEAD_TIME = 14

sales_df["average_lead_time"] = AVERAGE_LEAD_TIME
sales_df["maximum_lead_time"] = MAXIMUM_LEAD_TIME

print("Average Lead Time:", AVERAGE_LEAD_TIME, "days")
print("Maximum Lead Time:", MAXIMUM_LEAD_TIME, "days")
```

Safety stock

For HOBBIES_1_001:

Average Daily Sales ≈ 0.3136
Maximum Daily Sales = 5
Average Lead Time = 7
Maximum Lead Time = 14

Safety Stock
= (5 × 14) - (0.3136 × 7)
= 70 - 2.1952
≈ 67.80 units

```python
sales_df["safety_stock"] = (
    (sales_df["maximum_daily_sales"] * sales_df["maximum_lead_time"])
    -
    (sales_df["average_daily_sales"] * sales_df["average_lead_time"])
)

sales_df[
    [
        "item_id",
        "store_id",
        "average_daily_sales",
        "maximum_daily_sales",
        "average_lead_time",
        "maximum_lead_time",
        "safety_stock"
    ]
].head(10)
```

Reorder Point=(Average Daily Sales×Average Lead Time)+Safety Stock

For HOBBIES_1_001:

Average Daily Sales ≈ 0.3136
Average Lead Time = 7
Safety Stock ≈ 67.80

Reorder Point
= (0.3136 × 7) + 67.80
≈ 70.00 units

```python
sales_df["reorder_point"] = (
    (sales_df["average_daily_sales"] * sales_df["average_lead_time"])
    + sales_df["safety_stock"]
)

sales_df[
    [
        "item_id",
        "store_id",
        "average_daily_sales",
        "average_lead_time",
        "safety_stock",
        "reorder_point"
    ]
].head(10)
```

```python
inventory_optimization = sales_df[
    [
        "item_id",
        "store_id",
        "average_daily_sales",
        "maximum_daily_sales",
        "average_lead_time",
        "maximum_lead_time",
        "safety_stock",
        "reorder_point"
    ]
].copy()

inventory_optimization.head(10)
```

```python
numeric_columns = [
    "average_daily_sales",
    "maximum_daily_sales",
    "average_lead_time",
    "maximum_lead_time",
    "safety_stock",
    "reorder_point"
]

inventory_optimization[numeric_columns] = (
    inventory_optimization[numeric_columns].round(2)
)

inventory_optimization.head(10)
```

```python
inventory_optimization.to_csv(
    "../data/processed/inventory_optimization.csv",
    index=False
)

print("Inventory Optimization file saved successfully!")
print("Shape:", inventory_optimization.shape)
```
